```python
import pandas as pd
import re

def clean_text(text):
    text = text.lower()
    text = re.sub(r"[^\w\s]", "", text)
    return text.strip()

def load_data(file_path):
    data = []
    with open(file_path, 'r', encoding='utf-8') as file:
        for line in file:
            parts = line.strip().split(";")
            if len(parts) == 2:
                text, label = parts
                data.append((clean_text(text), label))
    return pd.DataFrame(data, columns=["text", "label"])
```
