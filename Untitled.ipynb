```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r"[^\w\s]", "", text)
    return text.strip()
```

```python
def load_data(file_path):
    data = []
    with open(file_path, 'r', encoding='utf-8') as file:
        for line in file:
            parts = line.strip().split(";")
            if len(parts) == 2:
                text, label = parts
                data.append((clean_text(text), label))
    return pd.DataFrame(data, columns=["text", "label"])
```

```python
train_df = load_data("train.txt")
val_df = load_data("val.txt")
```

```python
vectorizer = TfidfVectorizer(max_features=5000)
X_train = vectorizer.fit_transform(train_df["text"])
X_val = vectorizer.transform(val_df["text"])
```

```python
encoder = LabelEncoder()
y_train = encoder.fit_transform(train_df["label"])
y_val = encoder.transform(val_df["label"])
```

```python
class_weights = compute_class_weight(class_weight='balanced',
                                     classes=np.unique(y_train),
                                     y=y_train)
```

```python
class_weight_dict = dict(enumerate(class_weights))
```

```python
model = XGBClassifier(
    objective='multi:softmax',
    num_class=len(encoder.classes_),
    eval_metric='mlogloss',
    max_depth=6,
    n_estimators=300,
    learning_rate=0.1,
    random_state=42
)

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_val)
print(classification_report(y_val, y_pred, target_names=encoder.classes_))
```

```python
pickle.dump(model, open("emotion_model.pkl", "wb"))
pickle.dump(vectorizer, open("vectorizer.pkl", "wb"))
pickle.dump(encoder, open("label_encoder.pkl", "wb"))
```
